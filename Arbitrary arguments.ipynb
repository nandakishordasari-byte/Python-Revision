```python
def total_salary(*args):
    print(args)
    print(sum(args))
    print(max(args))
total_salary(1000, 2000, 3000)
```

```python
# arbitrary keyword arguments
def employee(**kwargs):
    print(kwargs)
employee(name="john", age=30, salary=50000)    
```

```python
# Nested functions
def outer_function():
    print("this is outer function.")
    def inner_function():
        print("this is inner function.")
    inner_function()
outer_function()        
```

```python
# Recursion
def countdown(n):
    if n==0:
        return
    print(n)
    countdown(n-1)
countdown(5)    
```
