```python
arr=[10, 20, 30, 40, 50]
print(arr.count(20))
print(len(arr))
print(arr.append(60))
print(arr)
print(arr.insert(1, 15))
print(arr)
print(arr.extend([70, 80]))
print(arr)
arr.remove(20)
print(arr)
arr.pop()
print(arr)
arr.pop(1)
print(arr)
del arr[0]
print(arr)
arr.clear()
print(arr)
```

```python
arr = [40, 10, 30, 20]
arr.sort()
print(arr)
arr.sort(reverse=True)
print(arr)
print(max(arr))
arr.sort()
print(arr[-2])
print(min(arr))
print(sum(arr))
```

```python
arr = [1, 2, 2, 3, 3, 4]
result = list(set(arr))
print(result)
```

```python
arr = [1, 2, 3, 4, 5, 6]
even = [x for x in arr if x % 2 == 0]
odd = [x for x in arr if x % 2 != 0]
print(even)
print(odd)
```

```python
arr = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(arr[0][1])
for row in arr:
    for value in row:
        print(value)
```
