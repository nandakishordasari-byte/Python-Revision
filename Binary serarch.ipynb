```python
arr=[2,5,8,10,15,20,25,30]
target=10
left=0
right=len(arr)-1
while left<=right:
    mid=(left+right)//2
    if arr[mid]==target:
        print("found")
        break
    elif arr[mid]<target:
        left=mid+1
    else:
        right=mid-1
```

```python
# 1.Target is the first element
arr=[2,5,8,12,16]
T=2
left=0
right=len(arr)-1
while left<=right:
    mid=(left+right)//2
    if arr[mid]==T:
        print("found",T)
        break
    elif arr[mid]<T:
        left=mid+1
    else:
        right=mid-1
```

```python
#2.Target is the last element
arr=[2,5,8,12,16]
T=16
left=0
right=len(arr)-1
while left<=right:
    mid=(left+right)//2
    print("mid",mid)
    if arr[mid]==T:
        print("found",T)
        break
    elif arr[mid]<T:
        left=mid+1
    else:
        right=mid-1
```

```python
# 3. Target doesn't exist
arr=[2,5,8,12,16]
T=10
left=0
right=len(arr)-1
while left<=right:
    mid=(left+right)//2
    print("mid",mid)
    if arr[mid]==T:
        print("found",T)
        break
    elif arr[mid]<T:
        left=mid+1
    else:
        right=mid-1
        print("not found",T)
```
