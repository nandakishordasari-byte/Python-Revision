```python
'''Find whether 20 exists
1. Find whether any number is greater than 20(x > 20)
2. Find whether any even number exists(x % 2 == 0)
3. Find the first negative number(x < 0)'''
```

```python
# Find whether 20 exists 
arr=[20,10,2,40,3,4]
T=20
for i in arr:
    if i==T:
        print(f"found {i} at {arr.index(i)}")
```

```python
# Find whether any number is greater than 20(x > 20)
arr=[20,70,2,10,30,40]
for i in arr:
    if i>=20:
        print(i)
```

```python
# Find whether any number is less than 20(x > 20)
M=[12,34,213,90,87,67,20]
for i in M:
    if i<=20:
        print(i)
```

```python
# Find whether any even number exists(x % 2 == 0)
N=[12,25,34,45,56,65,87,90,78]
for i in N:
    if i%2==0:
        print(i)
```

```python
#  Find whether any even number exists(x % 2 != 0)
N=[12,25,34,45,56,65,87,90,78]
for i in N:
    if i%2!=0:
        print(i)
```

```python
# Find the first negative number(x < 0)
B=[21,34,-9,-8,34,-90]
for i in B:
    if i<=0:
        print(i)
```
