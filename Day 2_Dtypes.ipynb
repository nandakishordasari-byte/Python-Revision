```python
# string
'''String is a sequence of characters. In Python, strings are enclosed in either single quotes or double quotes. Strings can be manipulated using various methods and operations.'''
D="   Nanda kishor   "
print(D[4])
print(D[0:2])
print(D[0:5:2])
print(D+" "+"Royal")
print(D.replace("Nanda", "Royal"))
print(len(D))
print(D.upper())
print(D.lower())
print(D.title())
print(D.capitalize())
print(D.count("a"))
print(D.find("kishor"))
print(D.split())
print(D.strip())
print(D.lstrip())
print(D.rstrip())
print("$".join(D))
print("Nan" in D) # Boolean value
print("nan" in D) # Boolean value
```

```python
# Tuple is immutable
'''Tuple is a collection which is ordered and unchangeable. In Python tuples are written with round brackets.'''
T=(1,2,3,4,5)
print(T)
print(T[0])
print(T[1:4])
print(T[0:5:2])
print(T.count(2))
print(T.index(3))
print(len(T))
print()
```

```python
# set is mutable and unordered
'''set is a collection of unique elements. It is mutable and unordered.
It does not allow duplicate elements. It is defined using curly braces {} or the set() function'''

friends={"Nanda","Diwaker","Naveen","Madhu"}    
print(friends.add("Vishnu"))
print(friends)
friends.update(["Ajay","Sendhil"])
print(friends)
print(friends.update(["Naveen", "Naveen Kumar"]))
print(friends)
friends.pop()
print(friends)
print("Nanda" in friends) #Boolean value
print("Rahul" in friends) #Boolean value
```
