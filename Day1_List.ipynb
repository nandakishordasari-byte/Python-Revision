```python
L = list(input(map(list, input().split())))
print(L)
```

```python
# 1 Finding the maximum value in List ofL
print(max(L))
print("The maximum value in the L is:",max(L))
print(L[-1])
```

```python
# 2 Finding the minimum value in List of L
print(min(L))
print("The minimum value in the L is:",min(L))
print(L[0])
```

```python
# 3 Finding the second maximum value in List of L
L.sort()
print(L[-2])
```

```python
# 4 Finding the middle value in List of L
middle=len(L)//2
print("The middle value in the L is:",L[middle])
```

```python
# 5 Finding the numbers in L which are divisible by 2
L=[12,34,87,23,54]
for i in L:
    if i % 2 == 0:
        print(i) 
```
