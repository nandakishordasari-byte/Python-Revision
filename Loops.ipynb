```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=" ")
    print()
```

```python
num=1
for i in range(1,6):
    for j in range(i):
        print(num,end=" ")
        num+=1
    print()    
```

```python
for i in range(1,6):
    for j in range(i):
        print(chr(j+65),end=" ")
    print()    
```

```python
for i in range(1,6):
    for j in range(i):
        print(chr(j+97),end=" ")
    print()    
```

```python
num=5
for i in range(num):
    print(" "*(num-i-1),end=" ")
    print("*"*(2*i+1))
```
