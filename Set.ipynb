```python
#set
N={21,34,54,87,43,23,12}
print(max(N))
print(min(N))
N=list(N)# second max
N.sort() 
print(N[-2])
N.sort() # second min
print(N[-5])
print(sum(N))
print(len(N))
print(sum(N)/len(N))# average
print(N.count(34))# count of 34
print([i for i in N if i%2==0])#even numbers
print([i for i in N if i%2!=0])#odd numbers
print([i for i in N if i>-0])#positive numbers
print([i for i in N if i<0])#negative numbers
print({i:N.count(i) for i in N})# Frequency Count
print(set(N))# unique elements
```
