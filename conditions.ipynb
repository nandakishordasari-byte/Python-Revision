```python
# If condiction
age = 20
if age >= 18:
    print("You are eligible to vote.")
```

```python
#else
num = -5
if num >= 0:
    print("Positive")
else:
    print("Negative")
```

```python
#elif
marks = 75
if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```
