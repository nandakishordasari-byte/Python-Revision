```python
balance=5000
try:
    amount=int(input("Enter Number:"))
    if amount>0:
        if amount>balance:
            print("insufficint balance")
        balance-=amount
        print("Available Balance:",balance)
        print("Debited Amount:",amount)
except Exception:
    print("Enter Numbers only")
```

```python
class InvalidAgeError(Exception):
    pass

try:
    age = int(input("Enter your age: "))

    if age < 18:
        raise InvalidAgeError("Age must be above 18.")

    print("Eligible to vote.")

except InvalidAgeError as e:
    print(e)
```

```python
from datetime import datetime
def record_login(username:str,status:str)->None:
    timestamp = datetime.now()
    log_message = f"{timestamp}|username={username}|status={status}\n"
    with open("login_activaty.log", "a") as file:
        file.write(log_message)

record_login(input("Enter your Name: "), "Success")
record_login(input("Enter your Name: "), "Falied")
```
