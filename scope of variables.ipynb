```python
# Local Scope
def student_name():
    name ="NANDA KISHOR"
    print(name)
student_name()    
```

```python
# Global Scope
company = "Algonex IT Solutions"
def company_name():
    print(company)
company_name()
print(company)    
```

```python
# use Global Keyword
n=20
def display():
    global n
    n=n*2
    print(n)
display()    
```

```python
# Local and Global Together
course="python"
def student_details():
    name="Nanda"
    print(course)
    print(name)
student_details()
print(course)    
```

```python
# Enclosing Scope
h=90
def outer():
    g=80
    print(h)
    def inner():
        print(h)
        print(g)
    inner()
outer()    
```

```python
# Enclosing using nonlocal keyword
t=10
def outer():
    c=200
    print(c)
    def inner():
        nonlocal c
        c=300
        print(c)
    inner()
    print(c)#nonlocal used to change the value of c in outer function
outer()
print(t)        
```

```python
# Local,Global and Enclosing scope
name="Nanda"
def outer():
    name="Kishor"
    def inner():
        name="Dasari"
        print(name)
    inner()
    print(name)
outer()
print(name)        
```

```python
# scope useing real student example
institute="Algonex IT Solutions"
def register_student(Student_name):
    course="python"
    def display():
        batch="morning batch"
        print("Institute:",institute)
        print("Student Name:",Student_name)
        print("course:",course)
        print("Batch:",batch)
    display()
register_student("Nanda Kishor")        
```

```python
# else Block scope
try:
    num=int(input())
    print(100/num)
except:
    print("Error")
else:
    print("program executed successfully")    
```

```python
# Finally Block scope
try:
    num=int(input())
    print(100/num)
except:
    print("Error")
finally:
    print("program executed successfully")    
```
